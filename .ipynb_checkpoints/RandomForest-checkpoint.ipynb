This notebook uses Random Forest classification to classify individuals into groups based on their sleeping disorders and visualize this classifier. The aim is to see how Random Forest classification performs at predicting sleeping disorders given different features, evaluate the model and compare it to the Gaussian Naieve Bayes model from the last notebook.

```python
%matplotlib inline
import pandas as pd
import matplotlib. pyplot as plt
import matplotlib.patches as mp
plt.style.use("seaborn-v0_8-whitegrid")
import numpy as np

import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from matplotlib.colors import ListedColormap
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
```

# Random Forest

The below model trains a random forest classifier to predict three different sleeping disorders and visualize them using physical activity level and age. The data is split into training and test data in order to evaluate the model. 

```python
dataset = pd.read_csv("Sleep_Health_and_Lifestyle_Dataset.csv")
```

```python
# Fill any nan values with "None"
dataset["Sleep Disorder"] = dataset["Sleep Disorder"].fillna("None")

# Get features and column stack to get coordinates
activityLevel = dataset["Physical Activity Level"].values
age = dataset["Age"].values
coords = np.column_stack((activityLevel, age))

# Encode labels
le = LabelEncoder()
labels = le.fit_transform(dataset["Sleep Disorder"])

# Colour mapping for visualization
custom_cmap = ListedColormap([
    "red",   # 0 Insomnia
    "green",     # 1 None
    "blue"     # 2 Sleep apnea
])

# Data is split into training and testing samples, 30% of the data is used for testing and 70% is used for training
X_train, X_test, y_train, y_test = train_test_split(coords, labels, test_size=0.3, random_state=42)

# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make prediction
y_pred = model.predict(X_test)
```

```python
def visualize_classifier(model, X_train, y_train, X_test=None, y_pred=None, cmap=None, xlabel="Feature 1", ylabel="Feature 2", title="Classifier"):

    if cmap is None:
        cmap = ListedColormap(["red", "green", "blue"])

    fig, ax = plt.subplots(figsize=(8,6))
    
    # Training points = circles
    ax.scatter(X_train[:, 0], X_train[:, 1], c=y_train, s=50, cmap=cmap,
               edgecolor='k', alpha=0.5, label='Train')
    
    # Decision boundary
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 200),
                         np.linspace(ylim[0], ylim[1], 200))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, levels=len(np.unique(y_train)), cmap=cmap)
    
    # Predicted points = triangles
    ax.scatter(X_test[:, 0], X_test[:, 1], c=y_pred, s=30, edgecolor='k', marker='^', cmap=cmap, label='Test')

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    plt.show()
```

```python
visualize_classifier(model, X_train, y_train, X_test, y_pred, cmap=custom_cmap, xlabel="Physical Activity Level", ylabel="Age", title="Random Forest: Physical Activity Level vs Age")
```

In this graph, green areas are areas where the model thinks no sleeping disorder is likely, red areas are where it thinks insomnia is likely and blue areas correspond with sleep apnea. Circles correspond with points from the training set and triangles correspond with points from the test set and their predicted sleeping disorders appear in green for no disorder, red for insomnia and blue for sleep apnea. From looking at this graph it appears as though every point may have been predicted correctly however it is not fully clear as multiple points could exist on top of one another. It also appears from looking at the graph that using Random Forest Classification gives us a better prediction and more accurate classification than Gaussian Naieve Bayes when given the same features. 

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")

# Heat map
sns.heatmap(cm.T, square=True, annot=True, fmt='d', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_,
            cmap='Blues')
plt.xlabel('True label')
plt.ylabel('Predicted label');  
```

This confusion matrix tells a different story and reveals that while the predictions are very accurate they are more or less the same as the Gaussian Naieve Bayes predictions from before.

While it is not visualized the next Random Forest Classification takes in a lot more sleep, health and lifestyle features from the dataset in an attempt to improve the model further.

```python
# Get features and column stack to get coordinates
age = dataset["Age"].values
sleepDuration = dataset["Sleep Duration"].values
sleepQuality = dataset["Quality of Sleep"].values
physicalActivity = dataset["Physical Activity Level"].values
stressLevel = dataset["Stress Level"].values
heartRate = dataset["Heart Rate"].values
dailySteps = dataset["Daily Steps"].values
coords = np.column_stack((age, sleepDuration, sleepQuality, physicalActivity, stressLevel, heartRate, dailySteps))

# Encode labels
le = LabelEncoder()
labels = le.fit_transform(dataset["Sleep Disorder"])

# Colour mapping for visualization
custom_cmap = ListedColormap([
    "red",   # 0 Insomnia
    "green",     # 1 None
    "blue"     # 2 Sleep apnea
])

# Data is split into training and testing samples, 30% of the data is used for testing and 70% is used for training
X_train, X_test, y_train, y_test = train_test_split(coords, labels, test_size=0.3, random_state=42)

# Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make prediction
y_pred = model.predict(X_test)

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")

# Heat map
sns.heatmap(cm.T, square=True, annot=True, fmt='d', cbar=False,
            xticklabels=le.classes_, yticklabels=le.classes_,
            cmap='Blues')
plt.xlabel('True label')
plt.ylabel('Predicted label');  
```

This confusion matrix shows a very slight improvement in correct predictions for no sleeping disorder despite the inclusion of many more features.
